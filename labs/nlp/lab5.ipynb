# **Stemming in NLP**

**Stemming** is a text-normalization technique in Natural Language Processing (NLP) that reduces words to a common *stem* by removing prefixes, suffixes, or other morphological endings.

The objective is to make different forms of a word match during text processing, particularly in information retrieval, search engines, and text classification.

## **1. How does it work?**

Consider the following words:

| Word        | Stem    |
| ----------- | ------- |
| connect     | connect |
| connected   | connect |
| connecting  | connect |
| connection  | connect |
| connections | connect |

A stemming algorithm attempts to identify and remove morphological affixes to produce a shared stem.

Notice that the resulting stem does not necessarily have to be a valid dictionary word.

For example:

| Word       | Stem  |
| ---------- | ----- |
| studies    | studi |
| studying   | studi |
| studied    | studi |
| relational | relat |
| generously | gener |

This is one of the defining characteristics of stemming: **it operates primarily through morphological rules and patterns rather than understanding the meaning of words.**

## **2. Common stemming algorithms**

There are several stemming algorithms, each with different levels of complexity and linguistic sophistication.

### **Porter Stemmer**

One of the most well-known stemming algorithms, developed by Martin Porter in 1980.

It applies a sequence of suffix-stripping rules to English words.

For example:

- `caresses` → `caress`
- `ponies` → `poni`
- `agreed` → `agre`
- `relational` → `relat`

### **Snowball Stemmer**

An evolution of the Porter stemming algorithm.

It provides stemming algorithms for multiple languages, including English, French, German, Spanish, and others.

### **Lancaster Stemmer**

A more aggressive stemming algorithm. It tends to produce shorter stems than Porter, sometimes at the expense of accuracy.

## **3. Stemming vs. Lemmatization**

These two techniques are often confused, but they have an important distinction.

| Aspect               | Stemming                  | Lemmatization                          |
| -------------------- | ------------------------- | -------------------------------------- |
| Approach             | Rule-based suffix removal | Vocabulary and morphological analysis  |
| Output               | Stem                      | Dictionary lemma                       |
| Linguistic awareness | Limited                   | Greater                                |
| Speed                | Generally faster          | Generally slower                       |
| Accuracy             | Less precise              | Generally more linguistically accurate |

For example:

| Word        | Porter stem | Lemma      |
| ----------- | ----------- | ---------- |
| studies     | studi       | study      |
| running     | run         | run        |
| better      | better      | good       |
| connections | connect     | connection |

The last example is particularly interesting: stemming can group *connect* and *connections* under the same stem, whereas lemmatization preserves their distinction.

## **5. Why is stemming useful?**

Consider a search engine indexing documents.

Suppose a user searches for:

`connecting devices`

But the indexed documents contain:

- Connected devices
- Connection between devices
- Connections among devices

Without normalization, a simple keyword-matching system might fail to recognize these as related.

By applying stemming during both indexing and querying, the search engine can match words sharing a common stem.

This can improve:

- Information retrieval.
- Search relevance.
- Document classification.
- Text clustering.
- Vocabulary reduction in NLP pipelines.

## **6. Limitations**

Stemming isn't perfect. Two particularly important problems are:

**Overstemming:** Different words are reduced to the same stem even though they have different meanings.

**Understemming:** Words that should be grouped together are assigned different stems.

For instance, stemming is not inherently semantic. It doesn't understand that two words are synonyms, nor does it necessarily distinguish different meanings of the same word.

## **7. A useful conceptual distinction**

It helps to think of text normalization as operating at different linguistic levels:

**Surface forms**

The actual words encountered in text.

`connected`, `connecting`, `connections`

**Morphological normalization**

Stemming or lemmatization.

`connect`, `connection`

**Semantic normalization**

Identifying words or expressions with related meanings.

`connect`, `link`, `associate`, `join`

These are different problems. Stemming addresses morphological variation, whereas semantic normalization attempts to capture relationships in meaning.
