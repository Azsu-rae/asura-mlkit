```python
def smart_lower(text):
    lowers = []
    for w in re.findall(r"\w+", text):
        if len(w) >= 2 and len(w) <= 5:
            lowers.append(w)
        else:
            lowers.append(w.lower())
    return lowers
        
text = "The NLP course covers USA-based case studies."
smart_lower(text)
```

```python
import re
text = "NLP is very useful."
result = re.sub(r"\s+", " ", text).strip()
print(result)
```

```python
def normalize_whitespace(text):
    result = []
    for line in text.split("\n"):
        result.append(re.sub(r"\s+", " ", line).strip())
    return "\n".join(result)
```

```python
text = """
for every node in the graph:
    we shall do  the   result
        and obtain the    optimization!
"""
print(normalize_whitespace(text))
```
