**A caveat about this output:** many of the tags are wrong. For example, `'ain'` as VB, `"didn't"` as NN, `'i'` as JJ, `'hers'` as VBZ, and `"they're"` as NN are all errors. That's expected: the tagger relies on sentence context, and here each word was tagged in isolation, which is the likely cause (apostrophe forms and clitic fragments like `d`, `ll`, `m`, `s`, `t`, `re` also confuse it). If you want accurate tags, tag actual sentences rather than a word list.
