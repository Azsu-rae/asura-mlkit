```python
text = "The course has 20 hours and 12 exercises."
result = re.sub(r"\d+", "<NUM>", text)
print(result)
```

```python
text = "The exam is on 16/09/2026 (16-09-2026) at 14:30."
found = re.finditer(r"(\d{2})[/-](\d{2})[/-](\d{4})", text)
for m in found:
    print("start:", m.start(1), "end:", m.end(1))
    print("span:", m.span(1))
# print(re.findall(r"\d{2}:\d{2}", text))
```
